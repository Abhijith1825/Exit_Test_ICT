Data Preprocessing

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Select numeric columns
numeric_cols = data.select_dtypes(include=['float64', 'int64']).columns

# Set up subplot grid
plt.figure(figsize=(16, 12))

for i, col in enumerate(numeric_cols, 1):
    plt.subplot(5, 4, i)
    sns.histplot(data[col], kde=True, bins=30)
    plt.title(f'{col}')
    plt.tight_layout()

numeric_cols = data.select_dtypes(include=['float64', 'int64']).columns
plt.show()
```

```python
target = "Best_Shopping_Mode"

# Treat blank values as missing
data = data.replace(r"^\s*$", pd.NA, regex=True)

# Remove rows where the target is missing
data = data.dropna(subset=[target]).reset_index(drop=True)

# Numeric columns
numeric_columns = [
    "Age",
    "Review_Influence_Score",
    "Social_Ads_Influence_Score",
    "Price_Comparison_Frequency",
    "Discount_Importance_Score",
    "Likely_To_Buy_After_Ad_Score",
    "Impulse_Purchase_Frequency",
    "Online_Shopping_Satisfaction_Score",
    "Trust_In_Online_Reviews_Score",
    "Regret_After_Purchase_Frequency"
]

# Convert invalid text values to NaN
for column in numeric_columns:
    data[column] = pd.to_numeric(data[column], errors="coerce")

# Fill numeric missing values with the median
for column in numeric_columns:
    data[column] = data[column].fillna(data[column].median())

# Categorical columns, excluding the target
categorical_columns = data.select_dtypes(
    include=["object", "category", "string"]
).columns

categorical_columns = categorical_columns.drop(target)

# Fill categorical missing values with the mode
for column in categorical_columns:
    mode_value = data[column].mode(dropna=True)

    if not mode_value.empty:
        data[column] = data[column].fillna(mode_value.iloc[0])

# Check remaining missing values
print(data.isna().sum())
```

```python
for i in numeric_cols:
  plt.figure(figsize=(5,5))
  plt.boxplot(data[i])
  plt.title(i)
```

```python
cat_cols = data.select_dtypes(include=['object']).columns
print(cat_cols)
```

```python
# List of columns to frequency encode
cols_to_encode = ['Country_Region','Top_Spending_Category','Occupation_Status','Avg_Monthly_Spend_NonEssentials']

for col in cols_to_encode:
    freq = data[col].value_counts() / len(data)
    data[col] = data[col].map(freq)

data
```

```python
from sklearn.preprocessing import LabelEncoder

# List of columns to label encode
cols_to_encode = ['Gender','Monthly_Income_Range','Shop_Most_Where','Online_Shopping_Frequency','Preferred_Platform',
                  'Primary_Device','Follows_Brands_On_Social','Makes_Shopping_List','Return_Frequency','Preferred_Payment_Method',
                  'Uses_BNPL_Installments','Reason_Prefer_Online','Reason_Prefer_InStore','Best_Shopping_Mode']

# Instantiate the encoder
label_enc = LabelEncoder()

# Loop through each column and transform it
for col in cols_to_encode:
    data[col] = label_enc.fit_transform(data[col])

data
```

Task 4 : Machine Learning Model

```python
y = data['Best_Shopping_Mode']
x = data.drop(columns=['Best_Shopping_Mode','Response_ID','Timestamp'])

from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler

std_scalar = StandardScaler()

numeric_col = x_train.select_dtypes(include=['int64', 'float64']).columns

x_train[numeric_col] = std_scalar.fit_transform(x_train[numeric_col])
x_test[numeric_col] = std_scalar.transform(x_test[numeric_col])

x_train.head()
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, f1_score, classification_report
from sklearn.linear_model import LogisticRegression

models = {
    'Logistic Regression': LogisticRegression(max_iter=1000, class_weight='balanced'),
    'Decision Tree': DecisionTreeClassifier(random_state=42, class_weight='balanced'),
    'Random Forest': RandomForestClassifier(random_state=42, class_weight='balanced'),
    'Gradient Boosting': GradientBoostingClassifier(random_state=42),
    'SVM (RBF)': SVC(class_weight='balanced', random_state=42),
    'KNN': KNeighborsClassifier(),
    'Naive Bayes': GaussianNB(),
}

results = []
fitted_models = {}
for name, model in models.items():
    model.fit(x_train, y_train)
    preds = model.predict(x_test)
    results.append((name, accuracy_score(y_test, preds), f1_score(y_test, preds, average='macro')))
    fitted_models[name] = model

results_df = pd.DataFrame(results, columns=['Model', 'Accuracy', 'Macro F1']).sort_values('Macro F1', ascending=False)
print(results_df.to_string(index=False))
```

```python
best_name = results_df.iloc[0]['Model']
best_model = fitted_models[best_name]
print(f'\nBest model: {best_name}')
print(classification_report(y_test, best_model.predict(x_test)))
```

```python
from sklearn.metrics import ConfusionMatrixDisplay
import matplotlib.pyplot as plt

# Predictions from the best model
y_pred_best = best_model.predict(x_test)

# Display confusion matrix
ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_pred_best,
    cmap="Blues",
    values_format="d"
)

plt.title("Confusion Matrix - Best Model")
plt.xlabel("Predicted Label")
plt.ylabel("Actual Label")
plt.tight_layout()
plt.show()
```
